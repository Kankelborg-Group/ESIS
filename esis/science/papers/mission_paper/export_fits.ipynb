```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from esis.data import level_3, level_4
import pathlib
from esis.flight import l3_events
```

```python
event = l3_events.otherx

l4 = level_4.Level_4.from_pickle(event.mart_inverted_pickle_path)

l3 = level_3.Level_3.from_pickle(level_3.ov_final_path)
```

```python
import tarfile
from astropy.io import fits
path = pathlib.Path.cwd() / ('esis_level_4_fits/'+event.name)
path.mkdir(parents=True, exist_ok=True)


for i, t in enumerate(l3.time):
    hdr = l4.wcs_list[i].to_header()
    hdr['DATE-OBS'] = str(l3.time[i])

    hdul = fits.HDUList()
    hdul.append(fits.PrimaryHDU(l4.cube_list[i], hdr))
    data_path = path / ('ESIS_level4_'+event.name+f'_{i}'+'.fits')
    hdul.writeto(data_path)
```

```python
event = l3_events.perfectx
l4 = level_4.Level_4.from_pickle(event.mart_inverted_pickle_path)

path = pathlib.Path.cwd() / ('esis_level_4_fits/'+event.name)
path.mkdir(parents=True, exist_ok=True)
```

```python
for i, t in enumerate(l3.time):
    hdr = l4.wcs_list[i].to_header()
    hdr['DATE-OBS'] = str(l3.time[i])

    hdul = fits.HDUList()
    hdul.append(fits.PrimaryHDU(l4.cube_list[i], hdr))
    data_path = path / ('ESIS_level4_'+event.name+f'_{i}'+'.fits')
    hdul.writeto(data_path)
```

```python

output_file = path.parent.name + '.tar.gz'
filename = path.parents[1] / output_file
filename
```

```python
with tarfile.open(filename, "w:gz") as tar:
    tar.add(path.parent, arcname=path.parent.name)
```
